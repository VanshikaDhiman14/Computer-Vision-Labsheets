```python
import cv2
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("img2.jpg")

if image is None:
    print("Error: Image not found!")
else:
    # Display original image
    print("Original Image:")
    cv2_imshow(image)

    # 1. Resize the image
    resized_image = cv2.resize(image, (500, 400))

    print("Resized Image:")
    cv2_imshow(resized_image)

    # 2. Crop the image
    # Format: image[y1:y2, x1:x2]
    cropped_image = image[50:300, 100:400]

    print("Cropped Image:")
    cv2_imshow(cropped_image)

    # 3. Rotate the image by 90 degrees clockwise
    rotated_image = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)

    print("Rotated Image:")
    cv2_imshow(rotated_image)

    # Save the images
    cv2.imwrite("resized_image.jpg", resized_image)
    cv2.imwrite("cropped_image.jpg", cropped_image)
    cv2.imwrite("rotated_image.jpg", rotated_image)

    print("All images saved successfully!")
```

```python
# 5. Split and Merge Image Channels
import cv2
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("img2.jpg")

if image is None:
    print("Error: Image not found!")
else:
    # Display original image
    print("Original Image:")
    cv2_imshow(image)

    # Split the image into Blue, Green, and Red channels
    blue, green, red = cv2.split(image)

    # Display individual channels
    print("Blue Channel:")
    cv2_imshow(blue)

    print("Green Channel:")
    cv2_imshow(green)

    print("Red Channel:")
    cv2_imshow(red)

    # Merge the channels back together
    merged_image = cv2.merge([blue, green, red])

    # Display merged image
    print("Merged Image:")
    cv2_imshow(merged_image)

    # Save the merged image
    cv2.imwrite("merged_image.jpg", merged_image)

    print("Image channels merged and saved successfully!")
```

```python
# 6. Split Image by Size (Tiling)
import cv2
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("img2.jpg")

if image is None:
    print("Error: Image not found!")
else:
    # Display original image
    print("Original Image:")
    cv2_imshow(image)

    # Get image dimensions
    height, width = image.shape[:2]

    # Define tile size
    tile_height = 100
    tile_width = 100

    # Split image into tiles
    count = 1

    for y in range(0, height, tile_height):
        for x in range(0, width, tile_width):

            # Crop the tile
            tile = image[y:y + tile_height, x:x + tile_width]

            # Display the tile
            print("Tile", count)
            cv2_imshow(tile)

            # Save the tile
            cv2.imwrite(f"tile_{count}.jpg", tile)

            count += 1

    print("Image split into", count - 1, "tiles successfully!")
```

```python
# 7. Merge Image from Tiles
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# Read the original image
image = cv2.imread("img2.jpg")

if image is None:
    print("Error: Image not found!")
else:
    # Get image dimensions
    height, width = image.shape[:2]

    # Define tile size
    tile_height = 100
    tile_width = 100

    # Create an empty image for merging
    merged_image = np.zeros_like(image)

    # Read and place tiles
    count = 1

    for y in range(0, height, tile_height):
        for x in range(0, width, tile_width):

            # Read the tile
            tile = cv2.imread(f"tile_{count}.jpg")

            if tile is not None:
                # Get tile dimensions
                h, w = tile.shape[:2]

                # Place tile in its original position
                merged_image[y:y+h, x:x+w] = tile

            count += 1

    # Display the merged image
    print("Merged Image:")
    cv2_imshow(merged_image)

    # Save the merged image
    cv2.imwrite("merged_from_tiles.jpg", merged_image)

    print("Image merged successfully as merged_from_tiles.jpg")
```

```python
# 8. Split Image into Four Quarters
import cv2
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("img2.jpg")

if image is None:
    print("Error: Image not found!")
else:
    # Get image dimensions
    height, width = image.shape[:2]

    # Find the middle points
    mid_height = height // 2
    mid_width = width // 2

    # Split image into four quarters
    top_left = image[:mid_height, :mid_width]
    top_right = image[:mid_height, mid_width:]
    bottom_left = image[mid_height:, :mid_width]
    bottom_right = image[mid_height:, mid_width:]

    # Display the four quarters
    print("Top-Left Quarter:")
    cv2_imshow(top_left)

    print("Top-Right Quarter:")
    cv2_imshow(top_right)

    print("Bottom-Left Quarter:")
    cv2_imshow(bottom_left)

    print("Bottom-Right Quarter:")
    cv2_imshow(bottom_right)

    # Save the four quarters
    cv2.imwrite("top_left.jpg", top_left)
    cv2.imwrite("top_right.jpg", top_right)
    cv2.imwrite("bottom_left.jpg", bottom_left)
    cv2.imwrite("bottom_right.jpg", bottom_right)

    print("Image successfully divided into four quarters!")
```
