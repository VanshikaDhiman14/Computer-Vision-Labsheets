```python

import cv2
import os
from google.colab.patches import cv2_imshow

# Enter image path
image_path = input("Enter the image path: ")

# Load image
image = cv2.imread(image_path)

# Check if image was loaded successfully
if image is None:
    print("Error: Could not load the image.")

else:
    # Get dimensions
    height, width = image.shape[:2]

    # Get number of channels
    if len(image.shape) == 2:
        channels = 1
    else:
        channels = image.shape[2]

    # Get datatype
    datatype = image.dtype

    # Get file size
    file_size = os.path.getsize(image_path)
    file_size_kb = file_size / 1024

    # Display properties
    print("\n--- Image Properties ---")
    print("Dimensions:", width, "x", height)
    print("Number of Channels:", channels)
    print("Datatype:", datatype)
    print("File Size: {:.2f} KB".format(file_size_kb))

    # Display image
    cv2_imshow(image)
```

```python
# Create a program that reads a colour image, converts it to grayscale and saves both versions side by side
# for visual comparision.
import cv2
from google.colab.patches import cv2_imshow
#Image path
image_path=input("Enter the image path: ")
#read the color image
image=cv2.imread(image_path)

if image is None:
  print("Error in loading the image.")
else:
  gray_image=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
  # Converting grayscale image to 3 channels(needed so that
  # both images have the same number of channels)
  gray_3channel=cv2.cvtColor(gray_image,cv2.COLOR_GRAY2BGR)
  comparision=cv2.vconcat([image,gray_3channel])
  #Display Comparision
  print("Original Image and Grayscale Image")
  cv2_imshow(comparision)
  cv2.imwrite("color_grayscale_comparision.jpg",comparision)
  print("Comparision image saved successfully!")
```

```python
# Develop an image format converter that takes one input image and exports it simultanneusly
# as JPG,PNG, and BMP ,then compares the resulting file sizes.

import cv2
import os
from google.colab.patches import cv2_imshow
image_path=input("Enter image path : ")

#Reading the image
image=cv2.imread(image_path)
if image is None:
  print("Error in reading the image!")
else:
  #Output file names
  jpg_file="converted_image.jpg"
  png_file="converted_image.png"
  bmp_file="converted_image.bmp"

  #Saving in different formats
  cv2.imwrite(jpg_file,image)
  cv2.imwrite(png_file,image)
  cv2.imwrite(bmp_file,image)
  # Get file sizes in KB
  jpg_size = os.path.getsize(jpg_file) / 1024
  png_size = os.path.getsize(png_file) / 1024
  bmp_size = os.path.getsize(bmp_file) / 1024
  # Display file sizes
  print("\n--- File Size Comparison ---")
  print(f"JPG : {jpg_size:.2f} KB")
  print(f"PNG : {png_size:.2f} KB")
  print(f"BMP : {bmp_size:.2f} KB")

  # Find smallest and largest
  sizes = {
        "JPG": jpg_size,
        "PNG": png_size,
        "BMP": bmp_size
  }
  smallest = min(sizes, key=sizes.get)
  largest = max(sizes, key=sizes.get)

  print("\nSmallest file:", smallest)
  print("Largest file:", largest)

  # Display the original image
  print("\nOriginal Image:")
  cv2_imshow(image)

  print("\nAll three files have been created successfully.")
```
