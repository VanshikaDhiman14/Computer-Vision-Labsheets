```python

import cv2
import os
from google.colab.patches import cv2_imshow

# Enter image path
image_path = input("Enter the image path: ")

# Load image
image = cv2.imread(image_path)

# Check if image was loaded successfully
if image is None:
    print("Error: Could not load the image.")

else:
    # Get dimensions
    height, width = image.shape[:2]

    # Get number of channels
    if len(image.shape) == 2:
        channels = 1
    else:
        channels = image.shape[2]

    # Get datatype
    datatype = image.dtype

    # Get file size
    file_size = os.path.getsize(image_path)
    file_size_kb = file_size / 1024

    # Display properties
    print("\n--- Image Properties ---")
    print("Dimensions:", width, "x", height)
    print("Number of Channels:", channels)
    print("Datatype:", datatype)
    print("File Size: {:.2f} KB".format(file_size_kb))

    # Display image
    cv2_imshow(image)
```

```python
# Create a program that reads a colour image, converts it to grayscale and saves both versions side by side
# for visual comparision.
import cv2
from google.colab.patches import cv2_imshow
#Image path
image_path=input("Enter the image path: ")
#read the color image
image=cv2.imread(image_path)

if image is None:
  print("Error in loading the image.")
else:
  gray_image=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
  # Converting grayscale image to 3 channels(needed so that
  # both images have the same number of channels)
  gray_3channel=cv2.cvtColor(gray_image,cv2.COLOR_GRAY2BGR)
  comparision=cv2.vconcat([image,gray_3channel])
  #Display Comparision
  print("Original Image and Grayscale Image")
  cv2_imshow(comparision)
  cv2.imwrite("color_grayscale_comparision.jpg",comparision)
  print("Comparision image saved successfully!")
```

```python
# Develop an image format converter that takes one input image and exports it simultanneusly
# as JPG,PNG, and BMP ,then compares the resulting file sizes.

import cv2
import os
from google.colab.patches import cv2_imshow
image_path=input("Enter image path : ")

#Reading the image
image=cv2.imread(image_path)
if image is None:
  print("Error in reading the image!")
else:
  #Output file names
  jpg_file="converted_image.jpg"
  png_file="converted_image.png"
  bmp_file="converted_image.bmp"

  #Saving in different formats
  cv2.imwrite(jpg_file,image)
  cv2.imwrite(png_file,image)
  cv2.imwrite(bmp_file,image)
  # Get file sizes in KB
  jpg_size = os.path.getsize(jpg_file) / 1024
  png_size = os.path.getsize(png_file) / 1024
  bmp_size = os.path.getsize(bmp_file) / 1024
  # Display file sizes
  print("\n--- File Size Comparison ---")
  print(f"JPG : {jpg_size:.2f} KB")
  print(f"PNG : {png_size:.2f} KB")
  print(f"BMP : {bmp_size:.2f} KB")

  # Find smallest and largest
  sizes = {
        "JPG": jpg_size,
        "PNG": png_size,
        "BMP": bmp_size
  }
  smallest = min(sizes, key=sizes.get)
  largest = max(sizes, key=sizes.get)

  print("\nSmallest file:", smallest)
  print("Largest file:", largest)

  # Display the original image
  print("\nOriginal Image:")
  cv2_imshow(image)

  print("\nAll three files have been created successfully.")
```

```python
# Write a Python program to resize multiple images in a folder while preserving their aspect ratio.
from google.colab import drive

drive.mount('/content/drive')

input_folder = '/content/drive/MyDrive/input_images'
output_folder = '/content/drive/MyDrive/resized_images'

os.makedirs(input_folder, exist_ok=True)
os.makedirs(output_folder, exist_ok=True)
```

```python
# How can we resize all images to a fixed width while maintaining their aspect ratio?
target_width = 300

for filename in os.listdir(input_folder):
    if filename.lower().endswith(('.png', '.jpg', '.jpeg')):

        img_path = os.path.join(input_folder, filename)
        img = cv2.imread(img_path)

        if img is None:
            print("Could not read:", filename)
            continue

        original_height, original_width = img.shape[:2]

        ratio = target_width / original_width

        new_width = target_width
        new_height = int(original_height * ratio)

        resized_img = cv2.resize(
            img,
            (new_width, new_height),
            interpolation=cv2.INTER_AREA
        )

        output_filename = f"resized_{filename}"
        output_path = os.path.join(output_folder, output_filename)

        cv2.imwrite(output_path, resized_img)

        print(
            f"Resized {filename} from "
            f"{original_width}x{original_height} "
            f"to {new_width}x{new_height}"
        )
```

```python
# Write a Python program to split an image into its Red, Green, and Blue channels.
b, g, r = cv2.split(image)

print("Blue channel shape:", b.shape)
print("Green channel shape:", g.shape)
print("Red channel shape:", r.shape)
```

```python
import numpy as np
# How can individual Red, Green, and Blue channels be displayed as colored images?
zeros = np.zeros(image.shape[:2], dtype="uint8")

blue_channel_img = cv2.merge([b, zeros, zeros])
green_channel_img = cv2.merge([zeros, g, zeros])
red_channel_img = cv2.merge([zeros, zeros, r])
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(blue_channel_img, cv2.COLOR_BGR2RGB))
plt.title("Blue Channel")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(green_channel_img, cv2.COLOR_BGR2RGB))
plt.title("Green Channel")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(red_channel_img, cv2.COLOR_BGR2RGB))
plt.title("Red Channel")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Write a Python program to reconstruct the original image by merging its individual color channels.
reconstructed_image = cv2.merge([b, g, r])

print("Reconstructed image shape:", reconstructed_image.shape)

plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(reconstructed_image, cv2.COLOR_BGR2RGB))
plt.title("Reconstructed Image")
plt.axis("off")
plt.show()
```

```python
# How can we verify that the reconstructed image is exactly the same as the original image?
if np.array_equal(image, reconstructed_image):
    print(
        "Pixel-level accuracy verified: "
        "Original and reconstructed images are identical."
    )
else:
    print(
        "Warning: Original and reconstructed images "
        "are NOT identical."
    )
```

```python
# How can we additionally check whether there is any pixel difference?
difference = cv2.subtract(image, reconstructed_image)

b, g, r = cv2.split(difference)

if cv2.countNonZero(b) == 0 and \
   cv2.countNonZero(g) == 0 and \
   cv2.countNonZero(r) == 0:
    print("Difference image is all zeros, confirming pixel-level accuracy.")
else:
    print("Difference image contains non-zero pixels.")
```

```python
# Write a Python program to add a text watermark to an image.
watermarked_image = image.copy()

text = "Sample Watermark"
font = cv2.FONT_HERSHEY_SIMPLEX
font_scale = 1
font_thickness = 2
text_color = (255, 255, 255)

(text_width, text_height), baseline = cv2.getTextSize(
    text,
    font,
    font_scale,
    font_thickness
)

padding = 20

x_pos = watermarked_image.shape[1] - text_width - padding
y_pos = watermarked_image.shape[0] - padding

cv2.putText(
    watermarked_image,
    text,
    (x_pos, y_pos),
    font,
    font_scale,
    text_color,
    font_thickness,
    cv2.LINE_AA
)
```

```python
# How can the watermarked image be displayed?
plt.figure(figsize=(8, 8))

plt.imshow(
    cv2.cvtColor(watermarked_image, cv2.COLOR_BGR2RGB)
)

plt.title("Watermarked Image")
plt.axis("off")
plt.show()
```
